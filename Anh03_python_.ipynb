```python
#1.Tạo frame
import pandas as pd
du_lieu={
    'Họ và tên': [
        'Nguyễn Văn A', 'Trần Thị B', 'Lê Hoàng C', 'Phạm Minh D', 'Vũ Thu E',
        'Hoàng Văn F', 'Đỗ Thị G', 'Phan Văn H', 'Bùi Thị I', 'Ngô Minh K'
    ],
    'Điểm chuyên cần': [9.0, 8.5, 10.0, 9.5, 7.0, 8.0, 9.0, 8.5, 9.0, 8.0],
    'Điểm giữa kỳ': [7.5, 8.0, 6.5, 9.0, 7.0, 5.5, 8.5, 6.0, 4.5, 9.5],
    'Điểm cuối kỳ': [8.0, 7.5, 9.0, 8.5, 6.0, 5.0, 8.0, 7.0, 5.5, 9.0]
}

df_sinh_viên = pd.DataFrame(du_lieu)
df_sinh_viên.index=df_sinh_viên.index +1

#2. Tổng kết
df_sinh_viên['Tổng_kết'] = ((0.2*df_sinh_viên['Điểm chuyên cần'])+
(0.3*df_sinh_viên['Điểm giữa kỳ'])+(0.5*df_sinh_viên['Điểm cuối kỳ']))

#3.Xếp loại sinh viên.
df_sinh_viên['Xếp loại']=pd.cut(
 df_sinh_viên['Tổng_kết'],
 bins=[0,0.5,7.0,8.5,10],
 labels=['yếu','trung bình','khá','giỏi'] )
df_sinh_viên
```

```python
#4. Thống kê
diem_tb=df_sinh_viên['Tổng_kết'].mean()
print(f"-Điểm trung bình của lớp là:{diem_tb:.2f}")
sinh_viên_đạt=df_sinh_viên[df_sinh_viên['Tổng_kết'] >= 5].shape[0]
print(f"-Số sinh viên đạt là:{sinh_viên_đạt}")
sinh_viên_điểm_cao_nhất= df_sinh_viên[df_sinh_viên['Tổng_kết'] == df_sinh_viên['Tổng_kết'].max()]
print("-Sinh viên có số điểm cao nhất là:")
display(sinh_viên_điểm_cao_nhất)
sinh_viên_điểm_thấp_nhất=df_sinh_viên[df_sinh_viên['Tổng_kết']==df_sinh_viên['Tổng_kết'].min()]
print("-Sinh viên có số điểm thấp nhất là:")
display(sinh_viên_điểm_thấp_nhất)


```

```python
#5.Trực quan hóa
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))
cac_cot = plt.bar(df_sinh_viên['Họ và tên'],df_sinh_viên['Tổng_kết'],color='royalblue',width=0.6)
plt.bar_label(cac_cot, padding=3, fmt='%.2f')
plt.title("Biểu đồ điểm tổng kết của các sinh viên",fontsize=14,fontweight='bold')
plt.xlabel("Tên sinh viên",fontsize=12)
plt.ylabel("Điểm tổng kết",fontsize=12)
plt.ylim(0,11)
plt.xticks(rotation=30)
plt.show()
```
